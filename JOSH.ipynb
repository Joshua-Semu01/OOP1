```python
class student:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def introduce(self):
        print(f"My name is {self.name} and i am {self.age} years old")
student1=student("JOSHUA",22)
student2=student("DAMARIS",22)
student1.introduce()
print(student1.name)
print(student1.age)
        
```

```python
class phone:
    def __init__(self,brand,storage):
        self.brand=brand
        self.storage=storage
    def display_infor(self):
        print(f"brand:{self.brand},storage:{self.storage}")
    def make_call(self,number):
        print(f"calling {number} from {self.brand}")
    phone1 = phone("IPHONE",64)
    phone2 = phone("SAMSUNG",64)
    phone1.display_infor()
    phone1.make_call("0789781923")
    phone2.display_infor()
    phone2.make_call("0777292239")
    
    
```
