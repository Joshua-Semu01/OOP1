```python
class BodaBoda:
    def __init__(self, name, weight, NFUNFU):
        self.name = name
        self.weight = weight
        self.umfuufu = NFUNFU
        self.__money = 0   # Encapsulated attribute

    def ride(self):
        print(self.name, "is riding the boda boda.")

    def show_details(self):
        print("Name:", self.name)
        print("Weight:", self.weight)
        print("NFUNFU:", self.NFUNFU)

    def set_money(self, money):
        self.__money = money

    def get_money(self):
        return self.__money


# Creating an object
boda1 = BodaBoda("Musa", 70, 20)

boda1.show_details()
boda1.ride()

boda1.set_money(50000)
print("Money:", boda1.get_money())
```
