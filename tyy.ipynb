```python
class BankAccount:
    def __init__(self, owner_name, starting_balance):
        self.owner = owner_name
        self.balance = starting_balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited {amount}. New balance: {self.balance}")
        else:
            print("Invalid deposit amount")

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"Withdrew {amount}. New balance: {self.balance}")
        else:
            print("Insufficient funds")

account1 = BankAccount("Alice", 20)
print(account1.owner)
print(account1.balance)
account1.deposit(50)
account1.withdraw(10)
```

```python
# ENCAPSULATION
class Student:
    def __init__(self, name, gpa):
        self.__name = name
        self.__gpa = gpa

    @property
    def name(self):
        return self.__name

    @property
    def gpa(self):
        return self.__gpa

    @gpa.setter
    def gpa(self, value):
        if 0.0 <= value <= 5.0:
            self.__gpa = value
        else:
            raise ValueError("GPA must be between 0.0 and 5.0")


student1 = Student("Alice", 4.0)

print(student1.gpa)
print(student1.name)



student1.gpa = 4.5
print(student1.gpa)
```

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance
    def get_balance(self):
        return self.__balance
    def deposit(self,amount):
        if amount > 0:
          self.__balance += amount
        else:
          print("Invalid deposit amount")
    def withdraw(self,amount):
        if 0 < amount <= self.__balance:
          self.__balance -= amount
        else:
          print("Invalid withdrawal amount")
    def check_balance(self):
        return self.__balance
    
acc=BankAccount(1000)
acc.deposit(500)
acc.withdraw(200)
print(acc.check_balance())
```

```python
class BodaBoda:
    def __init__(self, name, weight, umfuufu):
        self.name = name
        self.weight = weight
        self.umfuufu = umfuufu
        self.__money = 0   # Encapsulated attribute

    def ride(self):
        print(self.name, "is riding the boda boda.")

    def show_details(self):
        print("Name:", self.name)
        print("Weight:", self.weight)
        print("Umfuufu:", self.umfuufu)

    def set_money(self, money):
        self.__money = money

    def get_money(self):
        return self.__money


# Creating an object
boda1 = BodaBoda("Musa", 70, 20)

boda1.show_details()
boda1.ride()

boda1.set_money(50000)
print("Money:", boda1.get_money())
```

```python
class MomoWallet:
    def __init__(self, owner_name, balance=0):
        self.owner_name = owner_name
        self._balance = balance
    @property
    def balance(self):
        return self._balance
    @balance.setter
    def balance(self, value):
        if value < 0:
            raise ValueError("Balance cannot be negative.")
        self._balance = value
          
semu_wallet = MomoWallet("Semu Joshua", 2000)
semu_wallet._balance = -200
print(semu_wallet.owner_name)
print(f"{semu_wallet.owner_name}'s wallet balance: {semu_wallet.balance}")
```
